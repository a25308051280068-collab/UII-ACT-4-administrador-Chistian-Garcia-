```python
# 1. Inicialización del diccionario vacío
# Aquí es donde se guardará toda la información que ingreses
diccionario_administrador = {}

def agregar_administrador():
    print("\n--- Agregar Nuevo Administrador ---")
    # Los inputs capturan los datos para el VALUE
    empresa = input("Empresa (Esta será la CLAVE/KEY): ")
    depto = input("Departamento: ")
    presupuesto = input("Presupuesto asignado: ")
    personal = input("Personal a cargo: ")
    metas = input("Metas trimestrales: ")
    erp = input("Software ERP: ")

    # Se crea el registro vinculando la KEY (empresa) con su VALUE (diccionario interno)
    diccionario_administrador[empresa] = {
        "departamento": depto,
        "presupuesto": presupuesto,
        "personal": personal,
        "metas": metas,
        "erp": erp
    }
    print(f"\n¡Registro de {empresa} agregado con éxito!")
    listar_administrador()

def listar_administrador():
    print("\n--- Lista de Administradores (CBTIS 128) ---")
    if not diccionario_administrador:
        print("El diccionario está vacío actualmente. Agregue datos en la opción 1.")
    else:
        # Encabezados de la tabla
        print(f"{'Empresa (KEY)':<15} | {'Depto':<15} | {'Presupuesto':<12} | {'Personal':<10} | {'Metas':<15} | {'ERP':<10}")
        print("-" * 95)
        # Recorrido obligatorio usando for clave, valor (key, value)
        for empresa, datos in diccionario_administrador.items():
            print(f"{empresa:<15} | {datos['departamento']:<15} | {datos['presupuesto']:<12} | {datos['personal']:<10} | {datos['metas']:<15} | {datos['erp']:<10}")

def buscar_administrador():
    print("\n--- Buscar Administrador ---")
    nombre = input("Ingrese el nombre de la empresa (KEY) a buscar: ")
    if nombre in diccionario_administrador:
        # Recuperamos el VALUE asociado a esa KEY
        valor = diccionario_administrador[nombre]
        print(f"\n[RESULTADO] Empresa: {nombre}")
        print(f"Datos encontrados: {valor}")
    else:
        print("\nError: Esa clave no existe en el sistema.")
    listar_administrador()

def borrar_administrador():
    print("\n--- Borrar Administrador ---")
    nombre = input("Ingrese el nombre de la empresa a eliminar: ")
    if nombre in diccionario_administrador:
        # Eliminamos la KEY y automáticamente su VALUE
        del diccionario_administrador[nombre]
        print(f"\nRegistro de {nombre} eliminado correctamente.")
    else:
        print("\nError: No se puede borrar una empresa que no existe.")
    listar_administrador()

def actualizar_administrador():
    print("\n--- Actualizar Administrador ---")
    nombre = input("Ingrese el nombre de la empresa que desea modificar: ")
    if nombre in diccionario_administrador:
        print(f"Modificando datos para: {nombre}")
        # Se sobreescriben los VALUES manteniendo la misma KEY
        depto = input("Nuevo Departamento: ")
        presupuesto = input("Nuevo Presupuesto: ")
        personal = input("Nuevo Personal: ")
        metas = input("Nuevas Metas: ")
        erp = input("Nuevo Software ERP: ")

        diccionario_administrador[nombre] = {
            "departamento": depto,
            "presupuesto": presupuesto,
            "personal": personal,
            "metas": metas,
            "erp": erp
        }
        print("\nDatos actualizados.")
    else:
        print("\nError: Empresa no encontrada.")
    listar_administrador()

# --- Bloque Principal (Menú) ---
while True:
    print("\n================================================")
    print("      SISTEMA DE CONTROL DE ADMINISTRADORES      ")
    print("   CBTIS 128 | Alumno: Christian Garcia Jiemenez ")
    print("================================================")
    print("1. Agregar administrador (KEY:VALUE)")
    print("2. Listar administradores")
    print("3. Buscar administrador")
    print("4. Borrar administrador")
    print("5. Actualizar administrador")
    print("6. Salir")

    opcion = input("\nSeleccione una opción (1-6): ")

    if opcion == "1":
        agregar_administrador()
    elif opcion == "2":
        listar_administrador()
    elif opcion == "3":
        buscar_administrador()
    elif opcion == "4":
        borrar_administrador()
    elif opcion == "5":
        actualizar_administrador()
    elif opcion == "6":
        print("Cerrando programa... ¡Hasta luego!")
        break
    else:
        print("Opción inválida. Por favor, marque un número del 1 al 6.")
        print("Programa realizado por Christian Garcia NL = 21")
```
