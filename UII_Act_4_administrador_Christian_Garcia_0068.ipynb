```python
# Inicialización del diccionario vacío
diccionario_administrador = {}

def agregar_administrador():
    print("\n--- Agregar Nuevo Administrador ---")
    empresa = input("Empresa: ")
    depto = input("Departamento: ")
    presupuesto = input("Presupuesto asignado: ")
    personal = input("Personal a cargo: ")
    metas = input("Metas trimestrales: ")
    erp = input("Software ERP: ")

    # Guardamos los datos en el diccionario principal
    diccionario_administrador[empresa] = {
        "departamento": depto,
        "presupuesto": presupuesto,
        "personal": personal,
        "metas": metas,
        "erp": erp
    }
    listar_administrador()

def listar_administrador():
    print("\n--- Lista de Administradores ---")
    if not diccionario_administrador:
        print("El diccionario está vacío.")
    else:
        # Encabezados de la tabla
        print(f"{'Empresa':<15} | {'Depto':<15} | {'Presupuesto':<12} | {'Personal':<10} | {'Metas':<15} | {'ERP':<10}")
        print("-" * 85)
        # Recorrido usando for key, value
        for empresa, datos in diccionario_administrador.items():
            print(f"{empresa:<15} | {datos['departamento']:<15} | {datos['presupuesto']:<12} | {datos['personal']:<10} | {datos['metas']:<15} | {datos['erp']:<10}")

def buscar_administrador():
    print("\n--- Buscar Administrador ---")
    nombre = input("Ingrese el nombre de la empresa a buscar: ")
    if nombre in diccionario_administrador:
        datos = diccionario_administrador[nombre]
        print(f"Información encontrada: {datos}")
    else:
        print("No se encontró esa empresa.")
    listar_administrador()

def borrar_administrador():
    print("\n--- Borrar Administrador ---")
    nombre = input("Ingrese el nombre de la empresa a eliminar: ")
    if nombre in diccionario_administrador:
        del diccionario_administrador[nombre]
        print(f"Registro de {nombre} eliminado.")
    else:
        print("La empresa no existe.")
    listar_administrador()

def actualizar_administrador():
    print("\n--- Actualizar Administrador ---")
    nombre = input("Ingrese el nombre de la empresa a modificar: ")
    if nombre in diccionario_administrador:
        print("Ingrese los nuevos datos:")
        depto = input("Nuevo Departamento: ")
        presupuesto = input("Nuevo Presupuesto: ")
        personal = input("Nuevo Personal: ")
        metas = input("Nuevas Metas: ")
        erp = input("Nuevo Software ERP: ")

        diccionario_administrador[nombre] = {
            "departamento": depto,
            "presupuesto": presupuesto,
            "personal": personal,
            "metas": metas,
            "erp": erp
        }
    else:
        print("Empresa no encontrada.")
    listar_administrador()

# Precarga de 4 datos iniciales para cumplir con el requerimiento
diccionario_administrador["TechSol"] = {"departamento": "IT", "presupuesto": "50000", "personal": "10", "metas": "Migración", "erp": "SAP"}
diccionario_administrador["AutoMex"] = {"departamento": "Ventas", "presupuesto": "30000", "personal": "5", "metas": "Vender 100", "erp": "Oracle"}
diccionario_administrador["BuildCo"] = {"departamento": "RRHH", "presupuesto": "15000", "personal": "3", "metas": "Contratar", "erp": "Odoo"}
diccionario_administrador["Foodies"] = {"departamento": "Logística", "presupuesto": "25000", "personal": "8", "metas": "Entregas", "erp": "Dynamics"}

# Cuerpo principal - Menú interactivo
while True:
    print("\n********** MENÚ DE CONTROL **********")
    print("1. Agregar administrador")
    print("2. Listar administradores")
    print("3. Buscar administrador")
    print("4. Borrar administrador")
    print("5. Actualizar administrador")
    print("6. Salir")

    opcion = input("Seleccione una opción: ")

    if opcion == "1":
        agregar_administrador()
    elif opcion == "2":
        listar_administrador()
    elif opcion == "3":
        buscar_administrador()
    elif opcion == "4":
        borrar_administrador()
    elif opcion == "5":
        actualizar_administrador()
    elif opcion == "6":
        print("Saliendo del programa...")
        break
    else:
        print("Opción no válida, intente de nuevo.")
```
